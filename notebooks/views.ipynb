```python
from ledwall_server.plugins.clock import Clock
from ledwall_server.plugins.effect import SolidColor
from ledwall_server.plugins.elements import (
    BarElement,
    IconElement,
    RectElement,
    TextElement,
)
from ledwall_server.plugins.screen import Screen
from ledwall_server.plugins.text import StaticText
from ledwall_server.renderer import Renderer
from picow_ws2812_devtools.visualizer import LedWallVisualizer
```

```python
viz = LedWallVisualizer(width=32, height=24)
```

## Snapshot a single plugin

```python
clock = Clock({"color": [0, 255, 0]}, 32, 24)
viz.snapshot(plugin=clock)
```

## Snapshot a composed scene (dashboard)

```python
renderer = Renderer(32, 24)
layers = [
    (SolidColor({"color": [255, 0, 0]}, 16, 16), 0, 0),
    (Clock({"color": [0, 255, 0]}, 16, 16), 16, 0),
    (StaticText({"text": "HELLO", "color": [255, 200, 0]}, 32, 8), 0, 16),
]
viz.snapshot(renderer=renderer, layers=layers)
```

## Elements — building blocks

Elements can be drawn directly onto a Screen for quick previews.

```python
import matplotlib.pyplot as plt

screen = Screen(
    "elements_demo",
    [
        RectElement(0, 0, 32, 24, color=(30, 30, 30)),
        IconElement(1, 1, 8, 8, icon="sun", color=(255, 200, 0)),
        TextElement(10, 1, 20, 8, text="22°", color=(255, 255, 255)),
        BarElement(1, 12, 30, 3, value=0.65, color=(0, 200, 0), bg=(50, 50, 50)),
        RectElement(1, 18, 10, 5, color=(0, 100, 255), fill=False),
    ],
)

fig, ax = plt.subplots()
ax.imshow(screen.render(32, 24), interpolation="nearest")
ax.axis("off")
```
